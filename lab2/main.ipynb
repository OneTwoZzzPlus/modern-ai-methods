### Задание 1. Токенизация текстовых данных

```python
from pprint import pprint
from textwrap import fill

from nltk.tokenize import WordPunctTokenizer, sent_tokenize, word_tokenize

text = """My classmates don't agree on where to eat: some want pizza, while others want burgers.
Everyone argues about pi: is it 3.14 or 3.1415?
Visit our university website at https://my.itmo.ru or email me at 467335@niuitmo.ru.
Kolobok met a fox and said, "You can't eat me — I'm on a roll!"
"""

print("sent_tokenize (по предложениям)")
print(f"Токенов: {len(sent_tokenize(text))}")
pprint(sent_tokenize(text))

print("\nsent_tokenize (по словам)")
print(f"Токенов: {len(word_tokenize(text))}")
print(fill("\t".join(word_tokenize(text)), width=100))

print("\nWordPunctTokenizer " + r"(\w+|[^\w\s]+)")
print(f"Токенов: {len(WordPunctTokenizer().tokenize(text))}")
print(fill("\t".join(WordPunctTokenizer().tokenize(text)), width=100))
```

### Задание 2. Преобразование слов в их базовые формы с помощью стемминга

```python
from nltk.stem.lancaster import LancasterStemmer
from nltk.stem.porter import PorterStemmer
from nltk.stem.snowball import SnowballStemmer

words = [
    'studies', 'studying', 'happiness', 'generously', 'running', 'runner', 'relational', 
    'nationalization', 'connection', 'connected', 'organization', 'organizing', 'wolves',
    'leaves', 'children', 'mice', 'geese', 'better', 'went', 'cars', 'playing', 'played',
    'friendship', 'quickly', 'beautifully'
]

porter_stemmed = [PorterStemmer().stem(word) for word in words]
lancaster_stemmed = [LancasterStemmer().stem(word) for word in words]
snowball_stemmed = [SnowballStemmer(language="english").stem(word) for word in words]

print(f"{'Original':<16} {'Porter':<16} {'Lancaster':<16} {'Snowball':<16}")
print("=" * (16 * 4 + 3))
for word, porter_w, lancaster_w, snowball_w in zip(words, porter_stemmed, lancaster_stemmed, snowball_stemmed):
    print(f"{word:<16} {porter_w:<16} {lancaster_w:<16} {snowball_w:<16}")
```

### Задание 3. Преобразование слов в их базовые формы с помощью лемматизации

```python
from nltk.stem import WordNetLemmatizer

wnl = WordNetLemmatizer()

n_lemmatized = [wnl.lemmatize(word, pos="n") for word in words]
v_lemmatized = [wnl.lemmatize(word, pos="v") for word in words]

print(f"{'Original':<16} {'n form':<16} {'v form':<16}")
print("=" * (16 * 3 + 2))
for word, n_form, v_form in zip(words, n_lemmatized, v_lemmatized):
    print(f"{word:<16} {n_form:<16} {v_form:<16}")
```

### Задание 4. Разбиение текстовых данных на информационные блоки

```python
from nltk.corpus import brown

N = 1000
M = 50

sample = brown.words()[:N]
chunks = [sample[i:i + M] for i in range(0, len(sample), M)]

print(f"Всего чанков: {len(chunks)}")

for chunk in chunks[:20]:
    print(" ".join(chunk)[:50])
```

### Задание 5. Извлечение частотности слов с помощью модели Bag of Words

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer

documents = [" ".join(chunk) for chunk in chunks]
index_names = [f"chunk_{i}" for i in range(len(documents))]

vectorizer = CountVectorizer()
matrix = vectorizer.fit_transform(documents)

df = pd.DataFrame(
    matrix.toarray(),
    index=index_names,
    columns=vectorizer.get_feature_names_out(),
)

print(f"Размер матрицы: {df.shape[0]} чанков × {df.shape[1]} слов")
display(df)
```

### Задание 6. Создание прогнозатора категорий

```python
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer, TfidfTransformer
from sklearn.naive_bayes import MultinomialNB

categories = [
    'sci.space',
    'rec.sport.hockey',
    'talk.politics.misc',
    'comp.graphics',
    'sci.med',
]

training_data = fetch_20newsgroups(
    subset='train', categories=categories, shuffle=True, random_state=42
)

count_vectorizer = CountVectorizer()
train_counts = count_vectorizer.fit_transform(training_data.data)

tfidf_transformer = TfidfTransformer()
train_tfidf = tfidf_transformer.fit_transform(train_counts)

input_data = [
    "The astronauts launched the shuttle into orbit around the planet.",
    "The team scored two goals in the third period and won the game.",
    "The senator announced a new bill about tax reform in congress.",
    "Doctors recommend a new treatment for patients with heart disease.",
    "The designer used a new algorithm to render realistic 3D shadows.",
    "The new system uses advanced technology to improve the performance of the team."
]

classifier = MultinomialNB().fit(train_tfidf, training_data.target)

input_counts = count_vectorizer.transform(input_data)
input_tfidf = tfidf_transformer.transform(input_counts)

predictions = classifier.predict(input_tfidf)
for sentence, label in zip(input_data, predictions):
    print(f"{training_data.target_names[label]:20} | {sentence}")
```

### Задание 7. Создание анализатора грамматических родов

```python
import random

from nltk import NaiveBayesClassifier
from nltk.classify import accuracy as nltk_accuracy
from nltk.corpus import names


def gender_features(word, N):
    return {'suffix': word[-N:].lower()}

male_list = [(name, 'male') for name in names.words('male.txt')]
female_list = [(name, 'female') for name in names.words('female.txt')]
data = male_list + female_list

random.seed(42)
random.shuffle(data)

test_names = ['Ivan', 'Mia', 'Thomas', 'Nina', 'Alex', 'Sam', 'Taylor', 'Jordan']
Ns = range(1, 6)

results = {}
predictions = {name: {} for name in test_names}

for N in Ns:
    featuresets = [(gender_features(n, N), g) for (n, g) in data]
    split = int(0.8 * len(featuresets))
    train_set, test_set = featuresets[:split], featuresets[split:]

    classifier = NaiveBayesClassifier.train(train_set)
    results[N] = nltk_accuracy(classifier, test_set)

    for name in test_names:
        predictions[name][N] = classifier.classify(gender_features(name, N))

col_w = 10
print(f"{'Name':<12}" + "".join(f"{'N=' + str(N):<{col_w}}" for N in Ns))
print("=" * (12 + col_w * len(Ns)))
for name in test_names:
    print(f"{name:<12}" + "".join(f"{predictions[name][N]:<{col_w}}" for N in Ns))

print("-" * (12 + col_w * len(Ns)))
print(f"{'accuracy':<12}" + "".join(f"{results[N]:<{col_w}.4f}" for N in Ns))

best_N = max(results, key=results.get)
print(f"\nОптимальное N: {best_N}")
```

### Задание 8. Создание сентимент-анализатора

```python
from nltk.classify import NaiveBayesClassifier
from nltk.classify.util import accuracy as nltk_accuracy
from nltk.corpus import movie_reviews

fileids_pos = movie_reviews.fileids('pos')
fileids_neg = movie_reviews.fileids('neg')

def extract_features(words):
    return {word: True for word in words}

features_pos = [(extract_features(movie_reviews.words(fileids=[f])), 'Positive') for f in fileids_pos]
features_neg = [(extract_features(movie_reviews.words(fileids=[f])), 'Negative') for f in fileids_neg]

num_pos = int(0.8 * len(features_pos))
num_neg = int(0.8 * len(features_neg))
features_train = features_pos[:num_pos] + features_neg[:num_neg]
features_test = features_pos[num_pos:] + features_neg[num_neg:]

classifier = NaiveBayesClassifier.train(features_train)
print(f"Accuracy: {nltk_accuracy(classifier, features_test):.4f}")

classifier.show_most_informative_features(20)

test_reviews = [
    "this film is brilliant and wonderful with excellent acting",
    "this movie is boring and terrible with awful acting",
    "the story is not good and the acting is not convincing",
    "the film has great music but a terrible ending",
]

for review in test_reviews:
    probs = classifier.prob_classify(extract_features(review.split()))
    label = probs.max()
    print(f"{label:9} (P+={probs.prob('Positive'):.3f}, P-={probs.prob('Negative'):.3f}) | {review}")
```

### Задание 9. Тематическое моделирование с использованием латентного размещения Дирихле

```python
from gensim import corpora, models
from nltk.corpus import stopwords
from nltk.stem.snowball import SnowballStemmer
from nltk.tokenize import RegexpTokenizer

lines = [
    "The Roman empire expanded very rapidly and it was the biggest empire in the world for a long time.",
    "An algebraic structure is a set with one or more finitary operations defined on it that satisfies a list of axioms.",
    "Renaissance started as a cultural movement in Italy in the Late Medieval period and later spread to the rest of Europe.",
    "The line of demarcation between prehistoric and historical times is crossed when people cease to live only in the present.",
    "Mathematicians seek out patterns and use them to formulate new conjectures.",
    "A notational symbol that represents a number is called a numeral in mathematics.",
    "The process of extracting the underlying essence of a mathematical concept is called abstraction.",
    "Historically, people have frequently waged wars against each other in order to expand their empires.",
    "Ancient history indicates that various outside influences have helped formulate the culture and traditions of Eastern Europe.",
    "Mappings between sets which preserve structures are of special interest in many fields of mathematics.",
]

tokenizer = RegexpTokenizer(r'\w+')
stemmer = SnowballStemmer('english')
stop_words = set(stopwords.words('english'))

tokens = []
for line in lines:
    words = [stemmer.stem(w) for w in tokenizer.tokenize(line.lower()) if w not in stop_words]
    tokens.append(words)

dict_tokens = corpora.Dictionary(tokens)
doc_term_mat = [dict_tokens.doc2bow(t) for t in tokens]

ldamodel = models.ldamodel.LdaModel(
    doc_term_mat, num_topics=2, id2word=dict_tokens, passes=25, random_state=42
)

for topic_id, topic in ldamodel.show_topics(num_topics=2, num_words=5, formatted=False):
    print(f"Тема {topic_id}:")
    for word, weight in topic:
        print(f"  {word:<9}: {weight:.4f}")
```
